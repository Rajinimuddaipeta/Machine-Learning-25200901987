K-MEAN:


```python
import numpy as np

# Dataset
data = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [8, 7],
    [9, 8],
    [10, 9]
])

# Number of clusters
k = 2

# Initial centroids
centroids = data[[0, 3]]

# K-Means
for _ in range(10):

    # Calculate distances
    distances = np.linalg.norm(
        data[:, None] - centroids,
        axis=2
    )

    # Assign clusters
    clusters = np.argmin(distances, axis=1)

    # Update centroids
    centroids = np.array([
        data[clusters == i].mean(axis=0)
        for i in range(k)
    ])

# Output
print("Clusters:", clusters + 1)
print("Centroids:")
print(centroids)
```

```python
import numpy as np
from sklearn.cluster import KMeans

data = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [8, 7],
    [9, 8],
    [10, 9]
])

kmeans = KMeans(n_clusters=2, random_state=0, n_init=10)
kmeans.fit(data)

print("Clusters:", kmeans.labels_ + 1)
print("Centroids:")
print(kmeans.cluster_centers_)
```

```python
import numpy as np
from sklearn.cluster import KMeans

# Dataset
data = np.array([
    [1, 2], [2, 3], [3, 2], [2, 1], [3, 3],
    [10, 10], [11, 11], [12, 10], [11, 9], [10, 11],
    [20, 2], [21, 3], [22, 2], [21, 1], [22, 3],
    [5, 20], [6, 21], [7, 20], [6, 19], [5, 21],
    [20, 20], [21, 21], [22, 20], [21, 19], [22, 21]
])

# K-Means with 5 clusters
kmeans = KMeans(n_clusters=5, random_state=0, n_init=10)

# Train the model
kmeans.fit(data)

# Output
print("Clusters:", kmeans.labels_ + 1)

print("Centroids:")
print(kmeans.cluster_centers_)
```

DBSCAN:


```python
import numpy as np
from sklearn.cluster import DBSCAN

# Dataset
data = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [8, 7],
    [9, 8],
    [10, 9],
    [20, 20]
])

# DBSCAN
dbscan = DBSCAN(eps=2, min_samples=2)
clusters = dbscan.fit_predict(data)

# Output
print("Clusters:", clusters)
```

```python
import numpy as np
from sklearn.cluster import DBSCAN

# Dataset
data = np.array([
    # Cluster 1
    [1, 1], [2, 2], [2, 1], [1, 2], [3, 2], [2, 3],

    # Cluster 2
    [10, 10], [11, 11], [10, 11], [11, 10], [12, 11], [11, 12],

    # Cluster 3
    [20, 1], [21, 2], [20, 2], [21, 1], [22, 2], [21, 3],

    # Cluster 4
    [5, 15], [6, 16], [5, 16], [6, 15], [7, 16], [6, 17],

    # Cluster 5
    [20, 15], [21, 16], [20, 16], [21, 15], [22, 16], [21, 17],

    # Noise points
    [40, 40], [50, 5], [35, 20], [45, 30], [30, 45]
])

# DBSCAN
model = DBSCAN(eps=1.5, min_samples=3)

# Clustering
labels = model.fit_predict(data)

print("Clusters:")
print(labels)

print("\nNoise Points:")
print(data[labels == -1])
```
